```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import shapiro
```

```python
x = np.random.normal(size=300)
y = np.random.uniform(size=300)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

x = np.random.normal(size=300)


# Q‑Q plot para X (normal)
plt.figure()
stats.probplot(x, dist="norm", plot=plt)
plt.title("Q‑Q plot de X (Normal)")
plt.xlabel("Cuantiles teóricos")
plt.ylabel("Cuantiles de la muestra")
plt.show()

```

```python
# Q‑Q plot para Y (uniforme)

y = np.random.uniform(size=300)


plt.figure()
stats.probplot(y, dist="norm", plot=plt)
plt.title("Q‑Q plot de Y (Uniforme)")
plt.xlabel("Cuantiles teóricos")
plt.ylabel("Cuantiles de la muestra")
plt.show()
```

```python
from scipy.stats import levene

# Tres grupos de ejemplo
grupo1 = [10, 12, 9, 11, 10]
grupo2 = [20, 22, 19, 21, 23]
grupo3 = [30, 31, 29, 28, 35]

# Test de Levene (por defecto usa la mediana)
stat, p = levene(grupo1, grupo2, grupo3)

print("Estadístico de Levene:", stat)
print("p-value:", p)

if p > 0.05:
    print("No se rechaza H₀: las varianzas parecen iguales.")
else:
    print("Se rechaza H₀: al menos una varianza es distinta.")
```

```python
from scipy.stats import levene
# define groups
group_1 = [14, 34, 16, 43, 45,
           36, 42, 43, 16, 27]
group_2 = [34, 36, 44, 18, 42,
           39, 16, 35, 15, 33]

# define alpha
alpha = 0.05
# now we pass the groups and center value
# from the following
# ('trimmed mean', 'mean', 'median')
w_stats, p_value = levene(group_1, group_2,
                          center='mean')

if p_value > alpha:
    print("We do not reject the null hypothesis")
else:
    print("Reject the Null Hypothesis")
```
