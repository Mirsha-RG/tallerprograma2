```python
import pandas as pd
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/it-ces/Datasets/refs/heads/main/reglas(apriori)-info.csv")
```

```python
df['mgu'] = df['price_consec'] - df['unitary_cost_consec']
```

```python
df['expected-return'] = df['support'] * (df['price_consec'] - df['unitary_cost_consec']) * 100
```

```python
df.sort_values(by=['lift', 'expected-return'], ascending=False, inplace=True)
```
