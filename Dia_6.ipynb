# el índice de masa corporal el IMC sigue una distribución normal con $\mu = 27.6$ y una $\sigma = 5.2$ cual es la probabilidad de que al sacar una muestra de 15 indivudos se obtenga un promedio mayor a 28.6,


* la probabilidad de sacar un promedio menor a 23.

* la probabilidad de sacar un promedio entre 24 y 28.


```python
mu = 27.6
sigma = 5.2
promedios = []
trials = 100000
for _ in range(trials):
  promedios.append(np.random.normal(loc = mu, scale = sigma, size  = 15).mean())
```

```python
(np.array(promedios) > 28.6).sum()/trials
```

```python
print((np.array(promedios) < 23).sum()/trials)
print(norm.cdf(23, loc = mu, scale = sigma/(15**0.5)))
```

```python
((np.array(promedios) > 24) & (np.array(promedios) < 28)).sum()/trials
```
