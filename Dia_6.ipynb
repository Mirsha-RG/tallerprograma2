```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
mu_ingreso = 70
population_size = 700000
population = np.random.normal(loc = mu_ingreso, scale = 15, size = population_size)
```

```python
plt.hist(population, bins = 30);
np.random.choice(population, 35).mean()
```

```python
medias = []
trials = 100000
for _ in range(trials):
  medias.append(np.random.choice(population, 35).mean())
```

```python
plt.hist(medias);
```
