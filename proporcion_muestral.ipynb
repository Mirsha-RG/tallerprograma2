```python
import pandas as pd
import numpy as np
from scipy.stats import norm

df = pd.read_csv('https://raw.githubusercontent.com/it-ces/Datasets/main/Plenas(cotizantes)-car%C3%A1tula-2020.csv')
```

```python
df = df.rename(columns={'Clasificación Industrial Internacional Uniforme Versión 4 A.C (CIIU)': 'CIIU'})
```

```python
construccion = df[(df['CIIU'].str.startswith('F'))]
#construccion
len (construccion)
```

```python
proporcion_muestral = len(construccion)/ len(df) #la proporción muestral funciona como el "promedio"
proporcion_muestral
```

```python
total_empresas = len(df)
total_empresas
```

```python

z = 1.96

error_std = np.sqrt((proporcion_muestral * (1 - proporcion_muestral)) / total_empresas)

inferior = proporcion_muestral - z * error_std
superior = proporcion_muestral + z * error_std

print(inferior)
print(superior)
```
