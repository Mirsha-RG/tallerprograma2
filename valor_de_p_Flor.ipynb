```python
#simulacion de la normal y explicacion de p
#tengo una poblacion de niños cuya altura es 140 y ds de 10. cual es la probabilidad de que uno mida mas de 155?
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

media=140
desviacion=10
incognita=155

x = np.linspace(100, 180, 1000) #genero 1000 puntos entre 100 y 180
y = norm.pdf(x, media, desviacion) #se distribuyen normalmente con media

plt.plot(x, y)
```

```python
# Calculamos el valor p (probabilidad de medir más de 155 cm)
valor_p = norm.sf(incognita, media, desviacion) #survival function, complemento de CDF
print(valor_p)
```

```python
# Graficamos la curva normal
plt.plot(x, y)

# Rellenamos el área a la derecha de 155 (valor de p)
x_relleno = np.linspace(incognita, 180, 1000)
y_relleno = norm.pdf(x_relleno, media, desviacion)
plt.fill_between(x_relleno, y_relleno, color='skyblue', alpha=0.5, label=f'Valor p = {valor_p:.4f}')

plt.axvline(incognita, color='red', linestyle='--')

plt.show()

```
