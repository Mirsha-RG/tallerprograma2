```python
trials = 1000
n = 60
lanzamientos = 120
counter = 0

for _ in range(trials):
    if np.random.choice([0, 1], lanzamientos).sum() == n:
        counter += 1

porcentaje = (counter / trials) * 100
print(f"Ocurrió {counter} veces ({porcentaje:.2f}%)")
```

```python
from scipy.stats import binom, binomtest
import numpy as np

n = 120       # Número de lanzamientos
k = 100       # Número de "éxitos" (caras)
p = 0.5       # Probabilidad de éxito (moneda justa)

prob = binom.pmf(k, n, p)  # Probability Mass Function (PMF)
print(f"La probabilidad de obtener 100 veces 'cara' en 120 lanzamientos es: {prob:.10f}")
```
