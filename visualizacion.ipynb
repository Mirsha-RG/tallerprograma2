```python
# Importar librerías necesarias
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
x = np.arange(0, 11) # array con valores de 0 a 10

# tu codigo
#d = v * t

velocidad = 25/10
tiempo = 25

distancia = velocidad * tiempo

y = velocidad * x

y

plt.plot(x,y, marker = 'o')
plt.xlabel('tiempo')
plt.ylabel('distancia')
```

```python
df_carreras = pd.DataFrame({
    'Carrera': ['Ingeniería', 'Medicina', 'Derecho', 'Psicología'],
    'Estudiantes': [120, 80, 60, 100]
})

plt.bar(df_carreras['Carrera'], df_carreras['Estudiantes'])


plt.xlabel('Carreras')
plt.ylabel('Número de estudiantes')
plt.title('Estudiantes por carrera')
```

```python

colores = ['royalblue', 'tomato', 'goldenrod', 'mediumseagreen']

plt.bar(df_carreras['Carrera'], df_carreras['Estudiantes'], color=colores)

plt.title('Cantidad de Estudiantes por Carrera Universitaria')
plt.xlabel('Carrera')
plt.ylabel('Número de Estudiantes')
plt.show()
```

```python
colores = ['royalblue', 'tomato', 'goldenrod', 'mediumseagreen']



plt.barh(df_carreras['Carrera'], df_carreras['Estudiantes'], color=colores)

plt.title('Cantidad de Estudiantes por Carrera Universitaria')
plt.xlabel('Carrera')
plt.ylabel('Número de Estudiantes')
plt.show()
```

```python
plt.pie(
    df_carreras['Estudiantes'],
    labels=df_carreras['Carrera'],
    autopct='%1.1f%%',
    colors=['royalblue', 'tomato', 'goldenrod', 'mediumseagreen'],
    startangle=90

)

# Título del gráfico
plt.title('Distribución porcentual de estudiantes por carrera universitaria')
```

```python
df_iris = pd.DataFrame({ # DataFrame con datos aleatorios
    'sepal_length': np.random.normal(5.8, 0.4, 150),
    'sepal_width': np.random.normal(3.0, 0.3, 150),
    'petal_length': np.random.normal(3.7, 1.5, 150),
    'petal_width': np.random.normal(1.2, 0.8, 150),
    'species': np.random.choice(['setosa', 'versicolor', 'virginica'], 150)
})

df_iris
```

```python
# tu código

sns.scatterplot(
    data=df_iris,
    x='sepal_length',
    y='sepal_width',
    hue='species',
    palette='Set2'

)

plt.title('Relación entre largo y ancho del sépalo según especie de flor')
plt.xlabel('Largo del sépalo')
plt.ylabel('Ancho del sépalo')
```

```python


sns.histplot(
    data=df_iris,
    x='petal_length',
    bins=20,
    color = 'royalblue'
)

plt.title('Distribucion largo del petalo')
plt.xlabel('Largo del pétalo')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# tu código
colores = ['royalblue', 'goldenrod', 'mediumseagreen']

sns.histplot(
    data=df_iris,
    x='petal_length',
    hue = 'species',
    bins=20,
    palette = colores,
    alpha = 0.7,
    kde = True
)

plt.title('Distribucion del petalo por especie')
plt.xlabel('Largo del petalo')
plt.ylabel('Frecuencia')
```

```python
df_iris

# Matriz de correlacion
correlacion = df_iris.corr(numeric_only=True)

# Heatmap
sns.heatmap(
    correlacion,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'

    )

plt.title('Correlaciones entre Variables')
plt.tight_layout()
```

```python
# tu código
plt.figure(figsize=(8, 6))

sns.boxplot(data= df_iris, x='species', y= 'petal_length', palette='Set2')

plt.title('Comparacion del largo del petalo')
plt.xlabel('Especie')
plt.ylabel('Largo del petalo')
```
