```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/it-ces/Datasets/main/Plenas(cotizantes)-car%C3%A1tula-2020.csv"
```

```python
df = pd.read_csv(url)
```

```python
df.shape
```

```python
df.columns
```

```python
def extract(PALABRA):
  return PALABRA[0]
```

```python
df['sector'] = df['Clasificación Industrial Internacional Uniforme Versión 4 A.C (CIIU)'].apply(extract)
```

```python
df['sector'].value_counts()
```

```python
df['sector']
```

```python
df['construct'] = np.where(df['sector']=='F',1, 0)
```

```python
firms_construc  =  df['construct'].sum()
```

```python
firms_construc
```

```python
from statsmodels.stats.proportion import proportion_confint
```

```python
proportion_confint(firms_construc, df.shape[0], 0.05) # librería
```

```python
p_hat  = firms_construc/df.shape[0]
```

```python
p_hat
```

```python
from scipy.stats import norm
alpha  = 0.05
Z = 1- alpha/2
Z_alpha = norm.ppf(Z)
Z_alpha
```

```python
SE = ((p_hat * (1 - p_hat))/df.shape[0])**0.5
sup = p_hat  + Z_alpha * SE
low = p_hat - Z_alpha * SE
low, sup
```


Con una confianza del 95% elmtotal de empresas del sector constructor esta entre el 8 y 11 por ciento, es decir de 100 empresas entre 8 y 11 son de construccion.
