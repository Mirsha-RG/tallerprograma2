Tengo una poblacion de licoreras, su produccion es normal con Mu de 57, sigma de 7.8, saco una muestra de 35 licoreras, calculo el promedio de su producción y construyo el intervalo de confianza, hago lo anterior 200 mil veces.
el 95% de los intervalos captura a Mu

```python
# Parámetros
media_poblacional = 56  # μ
sigma_poblacional = 7.8  # σ
n = 35  # tamaño de muestra
z = 1.96  # para 95% de confianza
trials = 10000  # número de experimentos

prom_muestra = []
contador = 0

for _ in range(trials):
    # 1. Generar muestra
    muestra = np.random.normal(loc=media_poblacional, scale=sigma_poblacional, size=n)

    # 2. Calcular media muestral
    media_muestral = np.mean(muestra)
    prom_muestra.append(media_muestral)

    # 3. Calcular margen de error (usamos n, no la muestra)
    margen_error = z * (sigma_poblacional / np.sqrt(n))

    # 4. Calcular intervalos
    inferior = media_muestral - margen_error
    superior = media_muestral + margen_error

    # 5. Contar si contiene a la media poblacional
    if inferior <= media_poblacional <= superior:
        contador += 1

# 6. Calcular proporción
resultado = contador / trials
print(f"Proporción de intervalos que contienen μ: {resultado:.4f}")
```

```python
import numpy as np
import math

mu = 57
sigma = 7.8
n = 35
repeticiones = 200000
contador = 0 # para hacer el total de veces que mi mu_muestral cae dentro del intervalo de confianza

for _ in range(repeticiones):
  # obtengo la muestra
  muestra = np.random.normal(loc=mu, scale=sigma, size=n)
  # obtengo la media de la muestra
  mu_muestral = np.mean(muestra)
  # obtengo el error estándar
  se = sigma / math.sqrt(n)
  # para un intervalo de confianza del 95% mi z vale 1.96
  # construyo la derecha e izquierda
  lim_inf = mu_muestral - 1.96 * se
  lim_sup = mu_muestral + 1.96 * se

  # compruebo si mi mu está dentro del mismo
  if lim_inf <= mu <= lim_sup:
    contador += 1

resultado = contador/repeticiones

print(f"Las veces que mi mu muestral está dentro de mi intervalo de confianza es de: {resultado:.2%}")
```

```python
import numpy as np

# 1. Parámetros conocidos
mu = 57           # Media real
sigma = 7.8       # Desviación estándar poblacional
n = 35            # Tamaño de muestra
z = 1.96          # Valor z para 95% de confianza
num_sim = 200_000 # Número de simulaciones

# 2. Simular 200,000 muestras de tamaño 35
# Creamos una matriz de tamaño (200000, 35) con datos aleatorios normales
muestras = np.random.normal(loc=mu, scale=sigma, size=(num_sim, n))
print (muestras)

# 3. Calcular la media muestral de cada fila
medias_muestrales = np.mean(muestras, axis=1)

# 4. Calcular el error estándar (sigma / sqrt(n))
error_estandar = sigma / np.sqrt(n)

# 5. Construir los límites inferior y superior del intervalo
limite_inferior = medias_muestrales - z * error_estandar
limite_superior = medias_muestrales + z * error_estandar

# 6. Ver cuántos intervalos contienen a la media real (mu)
#contiene_mu almacena un array de valores booleanos (True o False) para cada uno de los 200,000 intervalos generados.
#Con el operador & (and lógico) se combinan las dos condiciones:
#El valor mu debe estar mayor o igual al límite inferior Y menor o igual al límite superior, Solo si ambas condiciones se cumplen,
# el resultado será True, es decir: ese intervalo contiene a la media real.

contiene_mu = (limite_inferior <= mu) & (mu <= limite_superior)

# 7. Calcular el porcentaje de intervalos que sí contienen a mu
proporcion_capturan_mu = np.mean(contiene_mu)

# 8. Mostrar el resultado
print(f"De los {num_sim} intervalos construidos, el {proporcion_capturan_mu*100:.2f}% capturan a la media real.")
```
