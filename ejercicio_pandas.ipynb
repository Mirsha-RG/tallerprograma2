```python
import pandas as pd
import numpy as np
from scipy.stats import norm
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/it-ces/Datasets/main/TMR.csv', decimal= ',')
df
```

```python
anio_2023 = df[(df['Año'] == 2023)]
anio_2023
```

```python
media_2023 = anio_2023['TRM'].mean()
media_2023
```

```python
std_2023 = df['TRM'].std()
std_2023
```

```python
n_2023 = len(anio_2023)
Z = 1.96

error_std_2023 = std_2023 / np.sqrt(n_2023)

limite_inferior = media_2023 - (error_std_2023 * Z)
limite_superior = media_2023 + (error_std_2023 * Z)

print(limite_inferior, limite_superior)
```

```python
febrero23 = anio_2023[(df['Mes'] =='Febrero')]
febrero23
```

```python
def nivel_deconfianza(datos, Z):

  media = datos.mean()
  std = datos.std()
  n = len(datos)
  error_std = std / np.sqrt(n)
  limite_inferior = media - (error_std * Z)
  limite_superior = media + (error_std * Z)

  return limite_inferior, limite_superior
```

```python
print(nivel_deconfianza(anio_2023['TRM'], Z))
#FILTRADO2023
```

```python
print(nivel_deconfianza(febrero23['TRM'], Z))
#FILTRADO2023
```
